In Python/pandas, DataFrames are objects. If you do df = df_raw, both names point to the same DataFrame.

```python
time_cols = ["Charging Start Time", "Charging End Time"]   # a list of 2 column names
for col in time_cols:
    df[col] = pd.to_datetime(df[col], errors = "coerce")
```

This is a loop: each column name in time_cols, it replaces that column with a datetime-parsed version.
-   pd.to.datetime(...) tries to parse strings like "2023-05-01 14:30" into a real datetime type.
-   errors="coerce" is the "don't crash" mode:
    - If a value is invalid/uparseable, pandas sets it to NaT("Not a Time"), which is like NaN but for datetimes.

```python
# Check how many failed to parse 
df[time_cols].isna().sum()
```

Breakdown:

- df[time_cols] selects just those 2 columns 
- .isna() turns each cell into True/False depending on whether it's missing (NaT counts as missing)
- .sum() counts how many True values per column

```python
df["session_duration_h"] = (
    df["Charging End Time"] - df["Charging Start Time"]
).dt.total_seconds() / 3600
```

- Subtracting 2 datetime columns gives a timedelta 
- .dt.total_seconds() converts the timedelta into secs
