- **Working title:** Predicting EV charging demand and charging session characteristics using real-world charging data
- **Course:** Data Science (BSc project)
- **Dataset:** EV Charging Patterns dataset (Kaggle)
- **High-level goal:** 
  - Understand *what affects EV charging behaviour* (session duration, energy consumed, arrival time, etc.)
